```python
import numpy as np
import matplotlib.pyplot as plt
import networkx as nx
from src.graph_building import gabriel_graph, gen_random_path
from src.graph_plotting import plot_graph, plot_path
```

```python
# Creating a plot of a Gabriel graph with 500 random points in the (-1, 1) square
np.random.seed(2)
x, y = np.random.uniform(-1, 1, 500), np.random.uniform(-1, 1, 500)
graph = gabriel_graph(np.column_stack((x, y)))
fig, ax = plot_graph(graph, x, y)
plt.show()
```

```python
# Plotting some random paths on the graph
colors = ["red", "blue", "green", "orange", "purple"]

np.random.seed(0)

fig, ax = plot_graph(graph, x, y)
for i in range(5):
    path = gen_random_path(graph, np.random.randint(5, 30))
    ax = plot_path(ax, path, x, y, color=colors[i])

plt.show()
```
